```python
import pandas as pd
from math import pi, sqrt
```

# 1 - Càlcul de les àrees del gràfic original

```python
df = pd.read_csv("../data/global_births_badgraph_data.csv")
df["Area(px^2)"] = df["Horizontal(px)"] * df["Vertical(px)"] * pi
```

```python
df["Radius(px)"] = df["Area(px^2)"].apply(lambda x: sqrt(x))
df = df[["Continent", "Births(M)", "Radius(px)"]]
df
```

# 2 - Tractament de la font de les dades

```python
un_wpp24_df = pd.read_csv("../data/UN_PPP2024_Output_Births.csv")
un_wpp24_df.info()
```

```python
un_wpp24_df = un_wpp24_df.drop(["Index","Variant","Notes","ISO3 Alpha-code","ISO2 Alpha-code","SDMX code**"], axis=1)
un_wpp24_df.info()
un_wpp24_df
```

```python
for i in range(2100 - 2024 + 1):
    un_wpp24_df[f"{2024 + i}"] = un_wpp24_df[un_wpp24_df[f"{2024 + i}"] != "..."][f"{2024 + i}"].apply(lambda x: int(x.replace(" ", "")))

un_wpp24_df
```

```python
treated_df = un_wpp24_df[un_wpp24_df["Type"] == "Region"]
treated_df
```

```python
source_df = treated_df.drop(["Location code", "Type", "Parent code"], axis=1)
source_df
```

```python
source_df = source_df[source_df["Region"] != "Americas"]
```

# 3 - Unió de datasets (àrea bombolles + font oficial)

```python
change_name = {
    214: "Latam & Caribbean",
    268: "North America"
               }

for k, v in change_name.items():
    source_df.loc[k, "Region"] = v
```

```python
source_df.rename({"Region": "Continent"}, inplace=True, axis=1)
```

```python
source_df
```

```python
for col in source_df.columns:
    if col != "Continent":
        source_df[col] = source_df[col].astype(int)
```

```python
source_df.to_csv("../data/global_births_evolution.csv", index=False)
```
