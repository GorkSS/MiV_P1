```python
import pandas as pd
from math import pi, sqrt
```

# 2 - Tractament de la font de les dades

```python
un_wpp24_df = pd.read_csv("../data/UN_PPP2024_Output_Births.csv")
un_wpp24_df.info()
```

```python
un_wpp24_df = un_wpp24_df.drop(["Index","Variant","Notes","ISO3 Alpha-code","ISO2 Alpha-code","SDMX code**"], axis=1)
un_wpp24_df.info()
un_wpp24_df
```

```python
for i in range(2100 - 2024 + 1):
    un_wpp24_df[f"{2024 + i}"] = un_wpp24_df[un_wpp24_df[f"{2024 + i}"] != "..."][f"{2024 + i}"].apply(lambda x: int(x.replace(" ", "")))

un_wpp24_df
```

```python
region_df = un_wpp24_df[un_wpp24_df["Type"] == "Region"]
region_df
```

```python
source_df = region_df[["Region", "Location code"]]
source_df
```

```python
source_df = source_df[source_df["Region"] != "Americas"]
```

# 3 - Unió de datasets (àrea bombolles + font oficial)

```python
change_name = {
    214: "Latam & Caribbean",
    268: "North America"
               }

for k, v in change_name.items():
    source_df.loc[k, "Region"] = v
```

```python
source_df.rename({"Region": "Continent"}, inplace=True, axis=1)
```

```python
source_df
```

```python
for col in source_df.columns:
    if col != "Continent":
        source_df[col] = source_df[col].astype(int)
```

```python
subregion_df = un_wpp24_df[un_wpp24_df["Type"] == "Subregion"]
subregion_df = subregion_df[["Location code", "Parent code"]]
subregion_df
```

```python
subregion_df.loc[len(subregion_df)] = [905, 905]
subregion_df
```

```python
countries_df = un_wpp24_df[un_wpp24_df["Type"] == "Country/Area"]
countries_df = countries_df.drop(["Location code", "Type"], axis=1)
countries_df
```

```python
countries_df[countries_df["Region"] == "Canada"]
```

```python
subregion_df = subregion_df.rename({"Parent code": "Continent code"}, axis=1)
merge_df = subregion_df.merge(countries_df, left_on="Location code", right_on="Parent code")
merge_df = merge_df.drop(["Location code", "Parent code"], axis=1)
merge_df
```

```python
merge_df[merge_df["Region"] == "Canada"]
```

```python
final_df = merge_df.merge(source_df, left_on="Continent code", right_on="Location code")
final_df = final_df.drop(["Continent code", "Location code"], axis=1)
final_df
```

```python
final_df["Continent"].drop_duplicates()
```

```python
for col in final_df.columns:
    if col != "Region" and col != "Continent":
        final_df[col] = final_df[col].astype(int)
```

```python
final_df.to_csv("../data/global_births_countries.csv", index=False)
```
